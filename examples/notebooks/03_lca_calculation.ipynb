# 03: Life Cycle Assessment (LCA) Walkthrough
Loading LCA evaluation engines and assessing environmental indicators under ISO 14040/14044 compliance.

```python
import sys
from pathlib import Path

# Add project root to path
PROJECT_ROOT = Path(".").resolve().parents[1]
sys.path.append(str(PROJECT_ROOT))

from pgloop import LCAEngine, get_pathway

print("LCA Engine loaded!")
```

## Evaluate LCA Impacts
Let's instantiate the engine and calculate environmental footprints for the Cement Production pathway.

```python
lca_engine = LCAEngine()
pathway = get_pathway("PG-CementProd")
print(f"Evaluating: {pathway.name}")

# Forward calculation
lca_result = lca_engine.calculate(pathway, functional_unit_value=1.0)

# Display outputs
for category, val in lca_result.impacts.items():
    unit = lca_engine.get_indicator_unit(category) or "kg-eq"
    print(f"  - {category:25}: {val:.4f} {unit}")
```
