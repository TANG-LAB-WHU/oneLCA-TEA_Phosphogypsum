# 04: Techno-Economic Analysis (TEA) Deep Dive
Loading TEA evaluation engines, pricing co-products, and modeling NPV, IRR, and payback period.

```python
import sys
from pathlib import Path

# Add project root to path
PROJECT_ROOT = Path('.').resolve().parents[1]
sys.path.append(str(PROJECT_ROOT))

from pgloop import TEAEngine, get_pathway
from pgloop.utils.currency import format_currency
print("TEA Engine loaded!")
```

## Evaluate Economic Feasibility
Let's instantiate the engine and calculate cash flow indices for the Rare Earth Extraction pathway.

```python
tea_engine = TEAEngine(country="China")
pathway = get_pathway("PG-REEextract")
print(f"Evaluating: {pathway.name}")

# 1. Calculate conventional CLCC & societal SLCC cost metrics
tea_result = tea_engine.calculate(pathway, functional_unit_value=1.0)
print(f"  - Conventional Cost (CLCC): {format_currency(tea_result.clcc)} / t PG")
print(f"  - Societal Cost (SLCC)    : {format_currency(tea_result.slcc)} / t PG")

# 2. Calculate NPV cash flow
npv_result = tea_engine.calculate_npv(pathway)
print(f"  - Discounted NPV          : {format_currency(npv_result['npv'])}")
print(f"  - IRR                     : {npv_result['irr'] * 100:.2f}%")
print(f"  - Payback Period          : {npv_result['payback_years']:.1f} years")
```
