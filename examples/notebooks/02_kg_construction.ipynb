# 02: Knowledge Graph Construction
Loading multi-modal extraction pipelines and querying local Vector-Graph RAG databases.

```python
import sys
from pathlib import Path

# Add project root to path
PROJECT_ROOT = Path('.').resolve().parents[1]
sys.path.append(str(PROJECT_ROOT))

from pgloop.knowledge import LightRAGEngine, RAGANYTHING_AVAILABLE
print(f"RAGAnything Available: {RAGANYTHING_AVAILABLE}")
```

## Query the Local LightRAG Database
Let's check if the processed database files exist and query them for domain facts.

```python
lightrag_dir = PROJECT_ROOT / "data" / "processed" / "lightrag_db"
if lightrag_dir.exists():
    rag = LightRAGEngine(working_dir=lightrag_dir)
    try:
        # Perform a mixed-mode vector-graph query
        res = rag.query("What are the main impurities in raw phosphogypsum?", mode="mix")
        print("Answer:", res.answer)
    except Exception as exc:
        print(f"Database exists but query failed: {exc}")
    finally:
        if hasattr(rag, "close"):
            rag.close()
else:
    print(f"LightRAG database not found at: {lightrag_dir}. Please run scripts/build_knowledge_graph.py first.")
```
